```python
''' IQR (InterQuartile Range)
====================================================================================
USE:
1. Numerical features
2. Skewed distributions
3. Sensitive models
4. Small-moderate outlier presence
====================================================================================
DON'T USE:
1. Small datasets
2. Valid outliers
3. Tree-based models
4. Extremely-skewed data
5. Categorical features
```

```python
''' HANDLING EXTREME OUTLIERS
====================================================================================
           [ remove ] -> true errors
             [ clip ] -> outliers are clearly errors, remove noise
        [ transform ] -> heavily skewed distribution
      [ winsorizing ] -> small outliers, preserve scale & interpretability
[ tree-based models ] -> robustness /wo data modification
   [ impute/replace ] -> if removing is too aggressive
```

```python
''' HANDLING CLASS IMBALANCE
====================================================================================
1. Random Oversampler
- duplicates existing samples in the minority class randomly
USE: quick balancing /w no extra computation, categorical
====================================================================================
2. SMOTE (Synthetic Minority Over-sampling Technique)
- generates synthetic samples, creates new, slightly different data points
USE: reduce overfitting & introduce diversity
    smote = SMOTE(random_state=42)
    X_resampled, y_resampled = smote.fit_resample(X, y)
====================================================================================
```

```python
''' IMPUTATION (HANDLING MISSING VALUES)
====================================================================================
   [ mean ] -> |Skewness| < 0.5, normal distribution, no significant outliers, few missing values
 [ median ] -> |Skewness| ≥ 0.5, skewed numerical features, present outliers, lots of missing values, robustness
   [ mode ] -> categorical features
====================================================================================
   [ 0-5% ] -> impute
  [ 5-30% ] -> impute + missing flag (for tree-based models)
[ >30-50% ] -> consider dropping unless important
   [ >50% ] -> usually drop unless critical feature or engineerable
```

```python
''' APPLYING TRANSFORMATION
====================================================================================
[   |Skewness| < 0.5   ] -> NO: don't apply transformation
[ 0.5 ≤ |Skewness| ≤ 1 ] -> OPTIONAL: apply transformation if model performance is affected
[    |Skewness| > 1    ] -> RECOMMENDED: investigate/apply transformation
====================================================================================
         [ Left-skewed ] -> reflect, then transform
        [ Right-skewed ] -> transform
====================================================================================
                [ sqrt ] -> mild skewness correction
                 [ log ] -> skewness correction/scale reduction
             [ Box-Cox ] -> power transform (positive values only)
         [ Yeo-Johnson ] -> handles 0s & negatives
```

```python
''' FEATURE ENGINEERING TECHNIQUES
====================================================================================
* [ FEATURE COMBINATION ]
1. AGGREGATION
- create summary features by grouping data
- multi-row-per-entity dataset

2. PAIRWISE MULTIPLICATION
- capture interactions between two features

3. RATIO/DIVISION
- understand proportions or efficiencies between features
- efficiency, density, proportion, etc.

4. POLYNOMIAL
- adds squared/cubic or interaction terms for numeric features
- captures non-linear relationships
====================================================================================
* [ ENCODING ]
1. TARGET ENCODING
- replace category /w average category's target value
- use `smoothing` to avoid overfitting
- high cardinality categoricals (many unique values)

2. ONE-HOT ENCODING
- convert categorical features into binary columns
- small cardinality categoricals (few unique values)

3. FREQUENCY ENCODING
- replace each category /w frequency
- keeps ordinal relationship, efficient for large categories

4. LABEL ENCODING
- convert categorical labels to numeric values (beneficial for tree-based models)
- high cardinality categoricals (many unique values)
====================================================================================
* [ BINNING ]
1. BINNING
- grouping of continuous numerical values into intervals/bins for categorization
- for non-tree-based models

2. QUANTILE BINNING
- grouping of continuous numerical values into proportional intervals/bins based on percentiles
- for non-tree-based models
====================================================================================
* [ OTHERS ]
1. PSEUDOLABELING
- use model's confident predictions on unlabeled data as `pseudolabels` to augment training dataset & fine-tune the model
- use if small data or lots of unlabeled data

2. MISSING FLAG
- feature indiciating if missing value
```
